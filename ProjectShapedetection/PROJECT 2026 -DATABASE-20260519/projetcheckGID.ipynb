```python
folder = "FORM1"

total = 0
correct = 0

for filename in os.listdir(folder):
    
    if filename.endswith(".jpg") or filename.endswith(".png") or  filename.endswith(".jpeg"):
        
        path = os.path.join(folder, filename)
        
        # lire image
        im = ski.imread(path)
        
        # lire groupe + ID
        
        groupe, identifiant = readGrId.readGroupeID(im)
        
        # extraire vrai ID depuis le nom
        # EXAM_FORM1_62401.jpg → 62401
        true_id = filename.split("_")[-1].split(".")[0]
        
        # print(f"\n📄 {filename}")
        # print(f"→ détecté : {identifiant}")
        # print(f"→ attendu : {true_id}")
        
        total += 1
        
        if identifiant == true_id:
            # print("✅ bon")
            # print(f"\n📄 {filename}")
            correct += 1
        else:
            # print("❌ faux")
            print(f"\n📄 {filename}")
            pass

# ===============================
# ✅ RESULTAT FINAL
# ===============================

if total > 0:
    accuracy = (correct / total) * 100
    print("\n===============================")
    print(f"✅ Taux de réussite : {accuracy:.2f}% ({correct}/{total})")
else:
    print("Aucune image trouvée")
```
